```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import kneighbors_graph
import networkx as nx
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# 1. Load your data
df = pd.read_csv("/content/515voice_309features_LOP_DiabetesDataset_Preprocessed_age35-60_Withoutfilename.csv")
X = df.drop(columns=["label"]).values
y = df["label"].values

# 2. Standardize (put every feature on the same scale)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Build the k-NN graph: connect each person to their 8 closest matches
K = 8
adj = kneighbors_graph(X_scaled, n_neighbors=K, mode="connectivity",
                        metric="cosine", include_self=False)

# 4. Make it two-way (symmetric)
adj = adj.maximum(adj.T)
A = adj.toarray()

print("Number of people (nodes):", A.shape[0])
print("Number of connections (edges):", int(A.sum() // 2))
print("Average number of neighbors per person:", A.sum(axis=1).mean())

# 5. Draw it
G = nx.from_numpy_array(A)
pos = nx.spring_layout(G, seed=42)
colors = ["orange" if label == 1 else "steelblue" for label in y]

plt.figure(figsize=(8, 8))
nx.draw_networkx_edges(G, pos, alpha=0.15)
nx.draw_networkx_nodes(G, pos, node_size=30, node_color=colors)
plt.title("My own k-NN graph (blue=non-diabetic, orange=diabetic)")
plt.axis("off")
plt.savefig("/content/my_own_graph.png", dpi=150)
print("Saved graph picture!")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from sklearn.model_selection import train_test_split

torch.manual_seed(42)
np.random.seed(42)

# ---- Prepare the graph for the model ----
n_nodes = A.shape[0]
A_hat = A + np.eye(n_nodes, dtype=np.float32)          # let each person "see" their own data too
D_hat = np.diag(1.0 / np.sqrt(A_hat.sum(axis=1)))       # balances out people with lots vs few neighbors
A_norm = D_hat @ A_hat @ D_hat
A_norm_t = torch.tensor(A_norm, dtype=torch.float32)
X_t = torch.tensor(X_scaled, dtype=torch.float32)
y_t = torch.tensor(y, dtype=torch.long)

# ---- Split people into train/validation/test groups ----
idx_all = np.arange(n_nodes)
idx_train, idx_temp = train_test_split(idx_all, test_size=0.4, stratify=y, random_state=42)
idx_val, idx_test = train_test_split(idx_temp, test_size=0.5, stratify=y[idx_temp], random_state=42)

train_mask = torch.zeros(n_nodes, dtype=torch.bool); train_mask[idx_train] = True
val_mask = torch.zeros(n_nodes, dtype=torch.bool); val_mask[idx_val] = True
test_mask = torch.zeros(n_nodes, dtype=torch.bool); test_mask[idx_test] = True

class_counts = np.bincount(y[idx_train])
class_weights = torch.tensor(class_counts.sum() / (2.0 * class_counts), dtype=torch.float32)

# ---- The GCN model itself ----
class GCNLayer(nn.Module):
    def __init__(self, in_dim, out_dim):
        super().__init__()
        self.linear = nn.Linear(in_dim, out_dim)
    def forward(self, A_norm, X):
        return self.linear(A_norm @ X)

class GCN(nn.Module):
    def __init__(self, in_dim, hidden_dim, out_dim):
        super().__init__()
        self.gc1 = GCNLayer(in_dim, hidden_dim)
        self.gc2 = GCNLayer(hidden_dim, out_dim)
    def forward(self, A_norm, X):
        h = F.relu(self.gc1(A_norm, X))
        h = F.dropout(h, p=0.5, training=self.training)
        return self.gc2(A_norm, h)

model = GCN(in_dim=X_t.shape[1], hidden_dim=64, out_dim=2)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=5e-4)

# ---- Train it: 200 rounds of learning ----
train_acc_history, val_acc_history = [], []
for epoch in range(1, 201):
    model.train()
    optimizer.zero_grad()
    out = model(A_norm_t, X_t)
    loss = F.cross_entropy(out[train_mask], y_t[train_mask], weight=class_weights)
    loss.backward()
    optimizer.step()

    model.eval()
    with torch.no_grad():
        out = model(A_norm_t, X_t)
        train_acc = (out[train_mask].argmax(1) == y_t[train_mask]).float().mean().item()
        val_acc = (out[val_mask].argmax(1) == y_t[val_mask]).float().mean().item()
    train_acc_history.append(train_acc)
    val_acc_history.append(val_acc)

    if epoch % 20 == 0:
        print(f"Round {epoch}: train accuracy {train_acc:.2f}, validation accuracy {val_acc:.2f}")

# ---- Final test: how good is it on people it never saw? ----
model.eval()
with torch.no_grad():
    out = model(A_norm_t, X_t)
    preds = out.argmax(1).numpy()

test_acc = (preds[idx_test] == y[idx_test]).mean()
print(f"\nFINAL TEST ACCURACY (on unseen people): {test_acc:.2%}")

# ---- Plot the learning curve ----
plt.figure(figsize=(6, 4))
plt.plot(train_acc_history, label="Train accuracy")
plt.plot(val_acc_history, label="Validation accuracy")
plt.xlabel("Training round")
plt.ylabel("Accuracy")
plt.title("How the GCN learned over time")
plt.legend()
plt.savefig("/content/training_curve.png", dpi=150)
print("Saved training_curve.png")
```
